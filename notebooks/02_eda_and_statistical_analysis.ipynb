# 02 — EDA et analyse statistique

L'objectif n'est pas seulement de tracer des courbes, mais de vérifier les hypothèses de saisonnalité, intermittence et association promotionnelle.


```python
from pathlib import Path
import pandas as pd
import plotly.express as px

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
PANEL = ROOT / "data" / "processed" / "curated_panel.parquet"
STATS = ROOT / "outputs" / "statistical_analysis"
panel = pd.read_parquet(PANEL)
history = panel[~panel.is_test].copy()
```

```python
profiles = pd.read_csv(STATS / "series_profiles.csv")
display(profiles.groupby("demand_class").size().rename("series_count"))
px.scatter(profiles, x="adi", y="cv2", color="demand_class", hover_data=["store_nbr", "family"])
```

```python
promo = pd.read_csv(STATS / "promotion_effects.csv")
display(promo.sort_values("relative_uplift", ascending=False))
px.bar(promo, x="family", y="relative_uplift", title="Association promotionnelle par famille")
```

**Attention :** l'uplift observé n'est pas causal. Les promotions sont probablement ciblées sur certaines périodes ou familles.
